```python
import pandas as pd
file_path= "netflix_titles.csv"
df= pd.read_csv(file_path,encoding='latin1')
df
```

```python
print(df.duplicated())
```

```python
df.head()
```

```python
total_entries=len(df)
total_columns=len(df.columns)
data_types=df.dtypes

print("overview of data set")
print(f"total number of entries:{total_entries}")
print(f"columns:{total_columns}")
print("data types:")
print(data_types)
```

```python
#analyisis of statistical columns
type_distribution=df['TYPE'].value_counts(normalize=True)

top_directors=df['DIRECTOR'].value_counts().head(10)

top_countries=df['COUNTRY'].value_counts().head(10)

rating_distribution=df['RATING'].value_counts(normalize=True)

duration_stats=df.groupby('TYPE')['DURATION'].describe()

top_genres=df['LISTED_IN'].value_counts().head(10)

#display
print('\nTYPE:')
print('\nDirector-top 10:')
print('\nCountry-top 10:')
print('\nRatings:')
print('\nDuration:')
print('\nListed_in-top 10:')
```

```python
release_year_stats=df['RELEASE_YEAR'].describe()
print('Release Year Statistics')
print(release_year_stats)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
plt.figure(figsize=(8,4))
plt.hist(df['RELEASE_YEAR'],bins=30,color='skyblue',edgecolor='black')
plt.title('release year distribution')
plt.xlabel('RELEASE_YEAR')
plt.ylabel('Frequency')
plt.grid(True)
plt.show()
```

```python
import seaborn as sns
```

```python
plt.figure(figsize=(8,6))
sns.countplot(x='TYPE',data=df,palette='pastel')
plt.title('distribution of titles by type')
plt.xlabel('TYPE')
plt.ylabel('count')
plt.show()
```

```python
#density plot for release year
plt.figure(figsize=(10,6))
df['RELEASE_YEAR'].plot(kind='density',color='skyblue')
plt.title('density plot of release year')
plt.xlabel('release year')
plt.ylabel('density')
plt.grid(True)
plt.show()
```

```python
#visualizations
plt.figure(figsize=(8,6))
sns.countplot(x='TYPE', data=df ,palette='pastel')
plt.title('Distribution of titles by type')
plt.xlabel('TYPE')
plt.ylabel('Count')
plt.show()
```

```python
#type distribution via piechart 
plt.figure(figsize=(8,8))
tv_show_rating_counts= df[df['TYPE']=='TV Show']['RATING'].value_counts()
plt.pie(tv_show_rating_counts, labels=tv_show_rating_counts.index,autopct= '%1.1f%%',
       startangle=140,colors=sns.color_palette('pastel'))
plt.title('tv_show_rating_counts')
plt.show()
```

```python
#top directors barplot
plt.figure(figsize=(8,6))
top_directors.plot(kind='bar',color='skyblue')
plt.title('top 10 directors by number of titles')
plt.xlabel('director')
plt.ylabel('number of titles')
plt.xticks(rotation=45,ha='right')
plt.show()
```

```python
#top countries barplot
plt.figure(figsize=(8,6))
top_countries.plot(kind='bar',color='lightgreen')
plt.title('top 10 countries by number production of netflix titles')
plt.xlabel('country')
plt.ylabel('number of titles')
plt.xticks(rotation=45,ha='right')
plt.show()
```

```python
#top genres barplot
plt.figure(figsize=(8,6))
top_genres.plot(kind='bar',color='red')
plt.title=('top ten genres')
plt.xlabel=('genre')
plt.ylabel=('number of titles')
plt.xticks(rotation=45,ha='right')
plt.show()
```

```python
#visualizing genres against directors
tdm= df[df['DIRECTOR'].isin(top_directors.index)&(df['TYPE']=='Movie')]
dgc=tdm.groupby('DIRECTOR')['LISTED_IN'].value_counts().unstack().fillna(0)

plt.figure(figsize=(12,8))
dgc.plot(kind='bar',stacked=True,colormap='tab20')
plt.title= 'distribution of genres among top directors'
plt.xlabel=('director')
plt.yabel=('number of movies')
plt.xticks(rotation=45)
plt.legend(title='genre',bbox_to_anchor=(1.05,1),loc='upper left')
plt.show()
```

```python
#bar plot for tvshows duration count
plt.figure(figsize=(10,4))
tv_show_duration=df[df['TYPE']=='TV Show']['DURATION'].value_counts().head(10)
sns.barplot(x=tv_show_duration.values,
            y=tv_show_duration.index, palette='pastel')

plt.title=('top 10 tv show durations')
plt.xlabel=('number of tv shows')
plt.yabel=('Duration')
plt.show()
```

```python
plt.figure(figsize=(10,4))
tv_show_genres=df[df['TYPE']=='TV Show']['LISTED_IN'].value_counts().head(10)
sns.barplot(x=tv_show_genres.values,
            y=tv_show_genres.index, palette='pastel')

plt.title=('top 10 tv show durations')
plt.xlabel=('number of tv shows')
plt.yabel=('Duration')
plt.show()
```

```python
#barplot for release year
plt.figure(figsize=(12,8))
release_year_count= df['RELEASE_YEAR'].value_counts().sort_index()
sns.barplot(x=release_year_count.index,
           y=release_year_count.values,
           palette='pastel')
plt.xlabel=('Release year')
plt.ylabel=('number of titles')
plt.xticks(rotation=90)
plt.show()
```
